```python
####This creates new filtered label files without touching originals:
# leaves out coniferous labels
import os
import shutil

LABELS_DIR = r"F:\Masterthesis\dataset\merged\labels"
FILTERED_DIR = r"F:\Masterthesis\dataset\merged\labels_filtered"

os.makedirs(FILTERED_DIR, exist_ok=True)

# remap: 1->0 (deciduous_tree), 2->1 (shrub), drop 0 (coniferous)
remap = {1: 0, 2: 1}

kept = 0
dropped_lines = 0
empty_files = 0

for fname in os.listdir(LABELS_DIR):
    if not fname.endswith(".txt"):
        continue
    src = os.path.join(LABELS_DIR, fname)
    dst = os.path.join(FILTERED_DIR, fname)
    new_lines = []
    with open(src) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            parts = line.split()
            cls = int(parts[0])
            if cls in remap:
                parts[0] = str(remap[cls])
                new_lines.append(" ".join(parts))
            else:
                dropped_lines += 1
    with open(dst, "w") as f:
        f.write("\n".join(new_lines))
    if new_lines:
        kept += 1
    else:
        empty_files += 1

print(f"Label files written: {kept + empty_files}")
print(f"Files with target labels: {kept}")
print(f"Background-only files (coniferous-only tiles): {empty_files}")
print(f"Coniferous lines dropped: {dropped_lines}")
```
